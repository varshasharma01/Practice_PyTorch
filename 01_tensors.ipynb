<h5> Other Ways

```python
# using arange 

b = torch.arange(1,10,2)
b
```

```python
# using linspace
l = torch.linspace(1,10, 10)
print("10 equally dist between 1 to 10",l)

# jab hum 2 numbers ke beech example 10 equal distance number chahein 
# we can also get 10 equally dist  between 1 to 5

a = torch.linspace(1,5,10)
print("10 equally dist between 1 to 5",a)
```

```python
# eye is used for diagonal matrix
a = torch.eye(3)
a
```

```python
# full is used to tell the shape of the matrix and which item be filled in that complete matrix

a = torch.full((3,3), 2)
a
```

<h6> now if we want to give a shape like any of our previous tensor

```python
x = torch.tensor(([1,2,3],[4,5,2]))
x
```

```python
torch.zeros_like(x)
```

<p> similarly

```python
b =torch.ones_like(x)
b
```

<p> but it will give error if we use exactly with rand because rand gives us float so we need to define the data type first

```python
b =torch.rand_like(x, dtype=torch.float16)
b
```

<h4> Tensor Data Types

```python
# find type of data
x.dtype
```

```python
# want to make tensor of int type
t = torch.tensor([[1.2,2]], dtype=int)
t
```

```python
d = torch.tensor([[1.2,2]], dtype=torch.int32)
d
```

```python
d = torch.tensor([[1.2,2]], dtype=torch.float16)
d
```

<p> We can change the existing data type of a tensor
