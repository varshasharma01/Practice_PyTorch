```python
import torch
import numpy as np
import pandas as pd
import math
```

```python
x = torch.tensor(2.0, requires_grad=True)
x
```

```python
y = x**2
y
```

```python
def dy_dx(x):
  return 2*x
```

```python
dy_dx(3)
```

```python
y.backward()
```

```python
x.grad
```

```python
z = math.sin(y)
```

```python
z
```

```python
def dz_dx(x):
  return 2*x * (math.cos(x**2))
```

```python
dz_dx(3.0)
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = x**2
z = torch.sin(y)
```

```python
z.backward()
x.grad
```
